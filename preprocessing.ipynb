```python
import pandas as pd
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
```

```python
# Step 1: Download required NLTK packages
print("Downloading NLTK packages...")
nltk.download('stopwords', quiet=True)
nltk.download('wordnet', quiet=True)
nltk.download('punkt', quiet=True)
```

```python
# Step 2: Load the dataset using the correct folder path
file_path = '/workspaces/product_review_classification_system/amazon_fine_food_review/Reviews.csv'
print(f"Loading dataset from {file_path}...")
df = pd.read_csv(r"/workspaces/product_review_classification_system/amazon_fine_food_review/Reviews.csv")
print(f"Initial dataset shape: {df.shape}")
```

```python
# Step 3: Clean missing values and duplicates
# For the Amazon dataset, we care about 'Text' and 'Score'
df = df.dropna(subset=['Text', 'Score'])
df = df.drop_duplicates(subset=['Text'])
print(f"Shape after cleaning nulls and duplicates: {df.shape}")
```

```python
# Step 4: Create sentiment labels from the 'Score' column (1-5 stars)
def map_sentiment(score):
    if score <= 2:
        return 'Negative'
    elif score == 3:
        return 'Neutral'
    else:
        return 'Positive'

df['sentiment'] = df['Score'].apply(map_sentiment)

print("\nSentiment Class Distribution:")
print(df['sentiment'].value_counts())
```

```python
# Step 5: Set up text cleaning pipeline
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def clean_text(text):
    if not isinstance(text, str):
        return ""
    
    # Lowercase
    t = text.lower()
    # Remove HTML tags, URLs, and numbers/punctuation
    t = re.sub(r'<.*?>', '', t)
    t = re.sub(r'https?://\S+|www\.\S+', '', t)
    t = re.sub(r'[^a-z\s]', '', t)
    
    # Tokenize, remove stop words, and lemmatize
    tokens = t.split()
    cleaned_tokens = [lemmatizer.lemmatize(w) for w in tokens if w not in stop_words]
    
    return " ".join(cleaned_tokens)

print("\nCleaning review texts (this may take a minute)...")
df['cleaned_text'] = df['Text'].apply(clean_text)
```

```python
# Step 6: Save the final output for Member 2
final_df = df[['cleaned_text', 'sentiment']]
output_file = 'preprocessed_reviews.csv'
final_df.to_csv(output_file, index=False)

print(f"\nSuccess! Cleaned dataset saved as '{output_file}'")
print(final_df.head(15))
```
